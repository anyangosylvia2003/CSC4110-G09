```python
# Inspect the column names and data types.
print("Column names:", df_raw.columns.tolist())
df_raw.info()

# Count missing values and duplicate rows.
print("\nMissing values by column:")
display(df_raw.isna().sum().to_frame(name="missing_count"))
print("Duplicate rows:", df_raw.duplicated().sum())

# Show summary statistics and counts for the Attrition target.
display(df_raw.describe(include="all"))
display(df_raw["Attrition"].value_counts(dropna=False).to_frame(name="count"))
```

```python
# Check how many unique values are in the columns expected to be constant.
constant_columns = ["EmployeeCount", "Over18", "StandardHours"]
display(df_raw[constant_columns].nunique(dropna=False).to_frame(name="unique_count"))

# Remove the constant columns and employee ID from a copy, keeping df_raw unchanged.
df_stage1 = df_raw.drop(columns=constant_columns + ["EmployeeNumber"]).copy()

print("df_stage1 shape:", df_stage1.shape)
print("Missing values:", int(df_stage1.isna().sum().sum()))
print("Duplicate rows:", int(df_stage1.duplicated().sum()))
```

```python
# Find any rows that break the staff-tenure rules.
company_violations = df_stage1["YearsAtCompany"] > df_stage1["TotalWorkingYears"]
role_violations = df_stage1["YearsInCurrentRole"] > df_stage1["YearsAtCompany"]

check_columns = ["YearsAtCompany", "TotalWorkingYears", "YearsInCurrentRole"]
print("Rows where YearsAtCompany exceeds TotalWorkingYears:", int(company_violations.sum()))
display(df_stage1.loc[company_violations, check_columns])

print("Rows where YearsInCurrentRole exceeds YearsAtCompany:", int(role_violations.sum()))
display(df_stage1.loc[role_violations, check_columns])

# Apply the IQR rule to YearsAtCompany and report outliers without changing them.
outlier_col = "YearsAtCompany"
q1 = df_stage1[outlier_col].quantile(0.25)
q3 = df_stage1[outlier_col].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outlier_mask = (df_stage1[outlier_col] < lower) | (df_stage1[outlier_col] > upper)

print(f"IQR bounds for {outlier_col}: {lower:.2f} to {upper:.2f}")
print("Number of IQR outliers:", int(outlier_mask.sum()))
display(df_stage1.loc[outlier_mask, check_columns].head(10))
```

```python
# Make a copy of the first eight distances and set the third value to missing.
import numpy as np

distance_demo = df_raw["DistanceFromHome"].head(8).copy()
distance_demo.iloc[2] = np.nan

# Sort the remaining values, find their median, and fill the missing value.
sorted_values = distance_demo.dropna().sort_values().tolist()
median_value = distance_demo.median()
filled_demo = distance_demo.fillna(median_value)

print("Sorted remaining values:", sorted_values)
print("Median replacement:", median_value)
print("Values after fillna:", filled_demo.tolist())
```

## Part b) Clean the data

1. Check and remove the constant columns and `EmployeeNumber`, keeping `df_raw` unchanged.
2. Check missing values, duplicates, and the two tenure rules; report any violating rows.
3. Apply the IQR rule to `YearsAtCompany`, check data types, and save the result as `df_stage1`.

```python
# Check constant columns, then drop them and EmployeeNumber from a copy.
constant_columns = ["EmployeeCount", "Over18", "StandardHours"]
display(df_raw[constant_columns].nunique(dropna=False).to_frame(name="unique_count"))

df_stage1 = df_raw.drop(columns=constant_columns + ["EmployeeNumber"]).copy()
print("df_stage1 shape:", df_stage1.shape)
print("Missing values:", int(df_stage1.isna().sum().sum()))
print("Duplicate rows:", int(df_stage1.duplicated().sum()))

# Check tenure rules and show any rows that break them.
company_violations = df_raw["YearsAtCompany"] > df_raw["TotalWorkingYears"]
role_violations = df_raw["YearsInCurrentRole"] > df_raw["YearsAtCompany"]

print("YearsAtCompany exceeds TotalWorkingYears:", int(company_violations.sum()))
display(df_raw.loc[company_violations, [
    "EmployeeNumber", "YearsAtCompany", "TotalWorkingYears"
]])

print("YearsInCurrentRole exceeds YearsAtCompany:", int(role_violations.sum()))
display(df_raw.loc[role_violations, [
    "EmployeeNumber", "YearsInCurrentRole", "YearsAtCompany"
]])

# Calculate IQR bounds for YearsAtCompany; keep plausible outliers uncapped.
outlier_col = "YearsAtCompany"
q1 = df_stage1[outlier_col].quantile(0.25)
q3 = df_stage1[outlier_col].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outlier_mask = (df_stage1[outlier_col] < lower) | (df_stage1[outlier_col] > upper)

print(f"IQR bounds for {outlier_col}: {lower:.2f} to {upper:.2f}")
print("IQR outliers kept:", int(outlier_mask.sum()))

# Check the cleaned dataset's data types.
df_stage1.info()

# Save the cleaned dataset in the shared data folder.
from pathlib import Path
output_path = Path("/content/drive/MyDrive/CSC4110_G09/data/stage1_hr.csv")
df_stage1.to_csv(output_path, index=False)
print("Saved:", output_path.exists(), "| Shape:", df_stage1.shape)
```
